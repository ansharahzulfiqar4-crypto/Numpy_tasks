```python
# 4. Important Array Properties

import numpy as np

# Create a 2D NumPy array
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# Print the array
print("Array:")
print(data)

# ndim - Number of dimensions
print("\nNumber of dimensions:")
print(data.ndim)

# shape - Number of rows and columns
print("\nShape:")
print(data.shape)

# size - Total number of values
print("\nTotal size:")
print(data.size)

# dtype - Data type of values
print("\nData type:")
print(data.dtype)
```

```python
# 5. Indexing

import numpy as np

# Create a NumPy array
numbers = np.array([10, 20, 30, 40, 50])

# Positive Indexing
print("First value:", numbers[0])
print("Third value:", numbers[2])

# Negative Indexing
print("Last value:", numbers[-1])
print("Second last value:", numbers[-2])
```

```python
# 6. Indexing in a 2D Array

import numpy as np

# Create a 2D array
marks = np.array([
    [70, 80, 90],
    [60, 75, 85]
])

# Print the complete array
print("Marks:")
print(marks)

# Access values using row and column
print("First row, second column:")
print(marks[0, 1])

print("Second row, third column:")
print(marks[1, 2])

print("First row, first column:")
print(marks[0, 0])
```

```python
# 7. Slicing

import numpy as np

# Create a NumPy array
numbers = np.array([10, 20, 30, 40, 50])

# Print complete array
print("Original Array:")
print(numbers)

# Slicing
print("\nSlicing [1:4]:")
print(numbers[1:4])

# More examples
print("\nSlicing [0:3]:")
print(numbers[0:3])

print("\nSlicing [2:5]:")
print(numbers[2:5])

print("\nSlicing [:3]:")
print(numbers[:3])

print("\nSlicing [2:]:")
print(numbers[2:])
```

```python
# 8. Vectorized Operations

import numpy as np

# Create a NumPy array
numbers = np.array([10, 20, 30, 40])

# Addition
print("Original Array:")
print(numbers)

print("\nAdd 5:")
print(numbers + 5)

# Multiplication
print("\nMultiply by 2:")
print(numbers * 2)

# Subtraction
print("\nSubtract 5:")
print(numbers - 5)

# Division
print("\nDivide by 2:")
print(numbers / 2)
```

```python
# 9. Operations Between Two Arrays

import numpy as np

# Create two NumPy arrays
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

# Print both arrays
print("Array a:")
print(a)

print("\nArray b:")
print(b)

# Addition
print("\nAddition:")
print(a + b)

# Subtraction
print("\nSubtraction:")
print(a - b)

# Multiplication
print("\nMultiplication:")
print(a * b)

# Division
print("\nDivision:")
print(a / b)
```

```python
# 10. Mathematical Functions

import numpy as np

# Square Root
numbers = np.array([4, 9, 16, 25])

print("Square Root:")
print(np.sqrt(numbers))


# Power
numbers = np.array([2, 3, 4])

print("\nPower:")
print(np.power(numbers, 2))


# Absolute Value
numbers = np.array([-10, -5, 4, 8])

print("\nAbsolute Value:")
print(np.abs(numbers))


# Round
numbers = np.array([4.2, 5.7, 8.9])

print("\nRound:")
print(np.round(numbers))
```

```python
# 11. Aggregation Functions

import numpy as np

# Create a NumPy array
marks = np.array([70, 85, 60, 95, 80])

# Sum
print("Sum:")
print(np.sum(marks))

# Average / Mean
print("\nAverage:")
print(np.mean(marks))

# Maximum
print("\nMaximum:")
print(np.max(marks))

# Minimum
print("\nMinimum:")
print(np.min(marks))

# Median
print("\nMedian:")
print(np.median(marks))

# Standard Deviation
print("\nStandard Deviation:")
print(np.std(marks))
```

```python
# 12. Conditions in NumPy

import numpy as np

# Create a NumPy array
marks = np.array([30, 50, 80, 45, 90, 20])

# Check which marks are greater than or equal to 50
print("Marks:")
print(marks)

print("\nMarks >= 50:")
print(marks >= 50)

# Other conditions
print("\nMarks > 50:")
print(marks > 50)

print("\nMarks < 50:")
print(marks < 50)

print("\nMarks == 50:")
print(marks == 50)
```

```python
# 13. Filtering Data

import numpy as np

# Create a NumPy array
marks = np.array([30, 50, 80, 45, 90, 20])

# Get passed marks
passed = marks[marks >= 50]

print("All Marks:")
print(marks)

print("\nPassed Marks:")
print(passed)

# Get failed marks
failed = marks[marks < 50]

print("\nFailed Marks:")
print(failed)
```

```python
# 14. Reshape

import numpy as np

# Create a 1D array
numbers = np.array([1, 2, 3, 4, 5, 6])

# Print original array
print("Original Array:")
print(numbers)

# Reshape into 2 rows and 3 columns
new_array = numbers.reshape(2, 3)

print("\nReshaped Array:")
print(new_array)
```

```python
# 15. Flatten

import numpy as np

# Create a 2D array
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Print original array
print("Original 2D Array:")
print(data)

# Convert 2D array into 1D array
flat_data = data.flatten()

print("\nFlattened 1D Array:")
print(flat_data)
```

```python
# 16. Useful NumPy Functions

import numpy as np

# zeros()
print("Zeros:")
print(np.zeros(5))

# ones()
print("\nOnes:")
print(np.ones(5))

# arange()
print("\nArange from 1 to 10:")
print(np.arange(1, 11))

# arange() with step
print("\nArange with step 2:")
print(np.arange(0, 11, 2))
```
